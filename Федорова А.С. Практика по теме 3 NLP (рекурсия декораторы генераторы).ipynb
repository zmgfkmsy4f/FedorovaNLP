## Задача 1. Базовый случай и шаг

*Новое: рекурсивный вызов*

**Рекурсия** - функция вызывает саму себя. В ней всегда две части:
**базовый случай** (ответ известен сразу, вызывать себя не нужно) и **шаг**
(вызов себя для задачи поменьше). Каждый шаг приближает к базовому случаю.

````python
def countdown(n):
    if n < 0:              # базовый случай
        print("старт!")
        return
    print(n)
    countdown(n - 1)       # шаг: задача на единицу меньше

countdown(3)
````

## Задача 2. Рекурсия возвращает результат

*Новое: «первый + остальные»*

Чаще рекурсивная функция **возвращает** ответ. Рецепт для списка:
ответ = первый элемент + ответ для остальных (`values[1:]`). Базовый случай -
пустой список: для него ответ известен сразу.

````python
def product(values):
    if len(values) == 0:                 # базовый случай
        return 1
    return values[0] * product(values[1:])

print(product([2, 3, 4]))                # 2 * (3 * (4 * 1)) = 24
````

## Задача 3. Цена рекурсии: Фибоначчи

*Новое: счётчик вызовов в словаре*

Числа Фибоначчи: каждое равно сумме двух предыдущих, `fib(0) = 0`, `fib(1) = 1`.
Рекурсивная запись короткая, но у неё есть цена. Чтобы её увидеть, заведём счётчик
вызовов. Храним его в **словаре**: функция меняет содержимое словаря, поэтому
изменение видно снаружи.

````python
CALLS = {"n": 0}

def add_up(n):
    """1 + 2 + ... + n рекурсией."""
    CALLS["n"] += 1
    if n == 0:
        return 0
    return n + add_up(n - 1)

print(add_up(10), " вызовов:", CALLS["n"])
````

# Часть 2. Декораторы

## Задача 4. Первый декоратор

*Новое: обёртка, `@`*

**Декоратор** - функция, которая принимает функцию и возвращает **новую**.
Новая функция («обёртка», `wrapper`) вызывает исходную и делает что-то до или после.
Запись `@loud` над `def` - то же самое, что `double = loud(double)`.

В шаблоне обёртки обязательны три вещи: `*args, **kwargs` в параметрах,
`func(*args, **kwargs)` внутри и `return` результата.

````python
def loud(func):
    def wrapper(*args, **kwargs):
        print("  до вызова", func.__name__)
        result = func(*args, **kwargs)
        print("  после вызова")
        return result
    return wrapper

@loud
def double(x):
    return x * 2

print(double(5))
````

**Ваша очередь:** напишите декоратор `announce`: перед вызовом печатает `вызываю` и имя функции,
после - `результат:` и то, что функция вернула, и возвращает этот результат.
Оберните с помощью `@` на новую функцию `first_word(text)`, которая возвращает
первое слово текста, и вызовите её для двух первых заголовков.

**Должно получиться:**

```

  результат: Сборная
  вызываю first_word
  результат: Тренер
```

````python
# Задача 4
def announce(func):
    def wrapper(*args, **kwargs):
        print("  вызываю", func.__name__)
        result = func(*args, **kwargs)
        print("  результат:", result)
        return result
    return wrapper
@announce
def first_word(text):
    return text.split()[0]
first_word(TITLES[0])
first_word(TITLES[1])
````

## Задача 5. Счётчик вызовов

*Новое: атрибут функции, декоратор на готовой функции*

У функции, как у любого объекта, можно создать **атрибут** - переменную,
прикреплённую к ней: `wrapper.last = None`. Обёртка меняет его при каждом вызове,
а прочитать его можно снаружи: `double.last`.

Декоратор можно надеть и на готовую функцию без `@`: `word_count = count_calls(word_count)`.
После этого **все**, кто вызывает `word_count` по имени, - в том числе другие ваши
функции - вызывают новую версию.

````python
def remember_last(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        wrapper.last = result        # атрибут функции: запоминаем результат
        return result
    wrapper.last = None
    return wrapper

@remember_last
def double(x):
    return x * 2

double(5)
double(21)
print("последний результат:", double.last)
````

## Задача 6. Кэш: Фибоначчи за 49 вызовов

*Новое: декоратор-кэш, два декоратора сразу*

**Кэш** - память ответов. Если функцию уже вызывали с такими аргументами,
обёртка возвращает запомненный ответ и не считает заново. Ключ словаря - кортеж
аргументов `args`, значение - результат.

Декораторов над функцией может быть несколько. Применяются они снизу вверх:
`@count_calls` над `@cache` значит `count_calls(cache(функция))`.

````python
def cache(func):
    def wrapper(*args):
        if args not in wrapper.memory:
            print("  считаю", args)
            wrapper.memory[args] = func(*args)
        return wrapper.memory[args]
    wrapper.memory = {}
    return wrapper

@cache
def square(x):
    return x * x

print(square(4), square(4), square(5))
````

# Часть 3. Генераторы

## Задача 7. Первый генератор

*Новое: `yield`, `next`*

Если в функции есть `yield`, это **генератор**. Вызов такой функции не выполняет
тело, а только создаёт объект-генератор. Каждый `next(gen)` выполняет тело
до следующего `yield`, отдаёт значение и замирает до следующего запроса.
Поэтому сообщения `print` внутри генератора появляются не сразу, а по одному.

````python
def numbers_loud():
    for i in [1, 2, 3]:
        print(f"  генератор: выдаю {i}")
        yield i

gen = numbers_loud()
print("генератор создан, но ещё ничего не выдал")
print("получил", next(gen))
print("получил", next(gen))
for rest in gen:                 # for забирает то, что осталось
    print("получил в цикле", rest)
````

## Задача 8. Генераторное выражение

*Новое: `(... for ...)`, одноразовость*

Короткий генератор записывают одной строкой - как списковое включение, но в
**круглых** скобках. Если выражение - единственный аргумент функции, вторые
скобки не пишут: `sum(x for x in xs)`.

Генератор - это **поток**, а не хранилище: прочитанное не вернуть, второй проход
пуст. Если значения нужны дважды, сохраните их в список.

````python
words = ["кот", "собака", "ёж", "попугай"]

lengths = (len(w) for w in words)
print("первая сумма:", sum(lengths))
print("вторая сумма:", sum(lengths))         # 0: уже прочитан

print("длинных слов:", sum(1 for w in words if len(w) > 3))
print("самое длинное:", max(len(w) for w in words))
````

## Задача 9. Цепочка генераторов

*Новое: генератор берёт значения у генератора*

Генераторы соединяют в цепочку: каждый берёт значения у предыдущего и передаёт
дальше. Значения идут по цепочке **по одному**, и если нужное найдено раньше,
остальное даже не читается. Так обрабатывают данные, которые не помещаются
в память.

````python
def read(items):
    for item in items:
        print("  читаю", item)
        yield item

def only_even(stream):
    for x in stream:
        if x % 2 == 0:
            yield x

for x in only_even(read([1, 2, 3, 4, 5, 6])):
    print("чётное:", x)
    if x == 4:
        break                    # 5 и 6 так и не прочитаны
````

**Ваша очередь:** напишите два генератора:

- `words(texts)` - выдаёт слова всех текстов по одному, в нижнем регистре
  и без знаков `.,:%` по краям (`word.strip(".,:%")`);
- `long_only(stream, min_len)` - пропускает дальше только слова длиной
  не меньше `min_len`.

Соедините их и выведите первые пять слов длиной от 12 букв (цикл с `break`).
Потом посчитайте, сколько всего таких слов в ленте (`sum(1 for ...)`).

**Должно получиться:**

```

региональных
аккумулятора
характеристик
всего слов от 12 букв: 16
```

````python
# Задача 9
def words(texts):
    for text in texts:
        for word in text.split():
            yield word.strip(".,:%").lower()
def long_only(stream, min_len):
    for word in stream:
        if len(word) >= min_len:
            yield word
shown = 0
for word in long_only(words(TEXTS), 12):
    print(word)
    shown += 1
    if shown == 5:
        break
print("всего слов от 12 букв:", sum(1 for _ in long_only(words(TEXTS), 12)))
````

## Шпаргалка занятия

| Что нужно | Как пишется |
| --- | --- |
| рекурсия | базовый случай + вызов себя для задачи поменьше |
| рекурсия над списком | `word_count(texts[0]) + total_words(texts[1:])` |
| счётчик снаружи | `CALLS = {"n": 0}`, внутри `CALLS["n"] += 1` |
| декоратор | `def deco(func): def wrapper(*args, **kwargs): ...; return wrapper` |
| внутри обёртки | `result = func(*args, **kwargs)`, `return result` |
| применить | `@deco` над `def` или `f = deco(f)` |
| атрибут функции | `wrapper.calls = 0`, снаружи `f.calls` |
| кэш | словарь `args -> результат` в `wrapper.memory` |
| генератор | функция с `yield`; `next(gen)` или `for` |
| генераторное выражение | `sum(1 for t in TEXTS if ...)` |
| читается один раз | нужен повторно - `list(...)` или создать заново |
| цепочка | `long_only(words(TEXTS), 12)` |

Подробнее - в конспекте 3.
